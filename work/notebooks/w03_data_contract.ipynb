## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Older content has much more historical data than newer content, so historical coverage is uneven.

Some early records may only include Google Search Console metrics, making them less informative than later records.

The dataset cannot account for external influences such as Google algorithm updates, competitor activity, or seasonal trends.

The data supports identifying patterns and making recommendations, but it can't prove that refreshing content will directly improve performance.
